```python
import pandas as pd
import numpy as np
import joblib
import os
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the saved Random Forest model and the feature names used in training
rf_model = joblib.load('models/water_contamination_rf_model.pkl')
meta = joblib.load('models/water_contamination_features.pkl')
feature_names = meta['features']

# 2. Build the importance table
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=True)

print(importance_df.sort_values('Importance', ascending=False).to_string(index=False))

# 3. Plot
plt.figure(figsize=(11, 7))
sns.barplot(data=importance_df, x='Importance', y='Feature', palette='magma', hue='Feature', legend=False)
plt.title('Random Forest - Water Feature Importance', fontsize=15, fontweight='bold')
plt.xlabel('Relative Importance Score', fontsize=13)
plt.ylabel('Predictor Features', fontsize=13)
plt.tight_layout()

os.makedirs('outputs', exist_ok=True)
plt.savefig('outputs/rf_feature_importance_water.png', dpi=300, bbox_inches='tight')
plt.show()
```
